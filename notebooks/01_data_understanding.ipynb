```python
from sklearn.model_selection import train_test_split

# Separate features and target
X = df_clean.drop("target", axis=1)
y = df_clean["target"]

# Split data: 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts(normalize=True))

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Numerical features
numeric_features = [
    "age",
    "resting_bp_s",
    "cholesterol",
    "max_heart_rate",
    "oldpeak"
]

# Categorical features
categorical_features = [
    "sex",
    "chest_pain_type",
    "fasting_blood_sugar",
    "resting_ecg",
    "exercise_angina",
    "ST_slope"
]

# Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

print("Numerical preprocessing pipeline created successfully.")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Categorical preprocessing
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine numerical and categorical preprocessing
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Complete preprocessing pipeline created successfully.")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# Create Logistic Regression pipeline
logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(max_iter=1000, random_state=42))
])

# Train the model
logistic_pipeline.fit(X_train, y_train)

# Make predictions
lr_pred = logistic_pipeline.predict(X_test)

print("Logistic Regression trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Evaluation metrics
accuracy = accuracy_score(y_test, lr_pred)
precision = precision_score(y_test, lr_pred)
recall = recall_score(y_test, lr_pred)
f1 = f1_score(y_test, lr_pred)
roc_auc = roc_auc_score(
    y_test,
    logistic_pipeline.predict_proba(X_test)[:, 1]
)

print("Logistic Regression Results")
print("-" * 35)
print(f"Accuracy : {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, lr_pred))
```

```python
cm = confusion_matrix(y_test, lr_pred)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["No Heart Disease", "Heart Disease"],
    yticklabels=["No Heart Disease", "Heart Disease"]
)

plt.title("Confusion Matrix - Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.tight_layout()
plt.show()
```

```python
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline

# Create Linear SVM pipeline
svm_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearSVC(
        dual="auto",
        random_state=42,
        max_iter=5000
    ))
])

# Train the model
svm_pipeline.fit(X_train, y_train)

# Make predictions
svm_pred = svm_pipeline.predict(X_test)

print("Linear SVM trained successfully.")
```

```python
# Evaluate Linear SVM

svm_accuracy = accuracy_score(y_test, svm_pred)
svm_precision = precision_score(y_test, svm_pred)
svm_recall = recall_score(y_test, svm_pred)
svm_f1 = f1_score(y_test, svm_pred)

print("Linear SVM Results")
print("-" * 30)
print(f"Accuracy : {svm_accuracy:.4f}")
print(f"Precision: {svm_precision:.4f}")
print(f"Recall   : {svm_recall:.4f}")
print(f"F1-Score : {svm_f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, svm_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Create Random Forest pipeline
rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        class_weight="balanced",
        n_jobs=-1
    ))
])

# Train the model
rf_pipeline.fit(X_train, y_train)

# Predictions
rf_pred = rf_pipeline.predict(X_test)

print("Random Forest trained successfully.")
```

# Model Performance Comparison

We compare all trained models using accuracy, precision, recall, F1-score, and ROC-AUC.
For heart disease prediction, recall and F1-score are particularly important because false negatives represent patients whose heart disease may be missed.

```python
from sklearn.metrics import roc_auc_score

# Logistic Regression
lr_auc = roc_auc_score(
    y_test,
    logistic_pipeline.predict_proba(X_test)[:, 1]
)

# Linear SVM
svm_auc = roc_auc_score(
    y_test,
    svm_pipeline.decision_function(X_test)
)

# Random Forest
rf_auc = roc_auc_score(
    y_test,
    rf_pipeline.predict_proba(X_test)[:, 1]
)

# Gradient Boosting
gb_auc = roc_auc_score(
    y_test,
    gb_pipeline.predict_proba(X_test)[:, 1]
)

# XGBoost
xgb_auc = roc_auc_score(
    y_test,
    xgb_pipeline.predict_proba(X_test)[:, 1]
)

print("ROC-AUC Scores")
print("-" * 30)
print(f"Logistic Regression : {lr_auc:.4f}")
print(f"Linear SVM          : {svm_auc:.4f}")
print(f"Random Forest       : {rf_auc:.4f}")
print(f"Gradient Boosting   : {gb_auc:.4f}")
print(f"XGBoost             : {xgb_auc:.4f}")
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Random Forest pipeline for tuning
rf_tune_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestClassifier(
        random_state=42,
        class_weight="balanced"
    ))
])

# Parameters to test
param_grid = {
    "model__n_estimators": [100, 200, 300],
    "model__max_depth": [None, 5, 10, 15],
    "model__min_samples_split": [2, 5, 10],
    "model__min_samples_leaf": [1, 2, 4]
}

# Grid Search
rf_grid = GridSearchCV(
    estimator=rf_tune_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="roc_auc",
    n_jobs=-1
)

# Train
rf_grid.fit(X_train, y_train)

print("Best Random Forest parameters:")
print(rf_grid.best_params_)

print("\nBest Cross-Validation ROC-AUC:")
print(f"{rf_grid.best_score_:.4f}")
```

```python
# Get predictions from the tuned Random Forest
rf_tuned_pred = rf_grid.predict(X_test)

# Get probabilities for ROC-AUC
rf_tuned_prob = rf_grid.predict_proba(X_test)[:, 1]

# Calculate metrics
rf_tuned_accuracy = accuracy_score(y_test, rf_tuned_pred)
rf_tuned_precision = precision_score(y_test, rf_tuned_pred)
rf_tuned_recall = recall_score(y_test, rf_tuned_pred)
rf_tuned_f1 = f1_score(y_test, rf_tuned_pred)
rf_tuned_auc = roc_auc_score(y_test, rf_tuned_prob)

print("Tuned Random Forest Results")
print("-" * 35)
print(f"Accuracy : {rf_tuned_accuracy:.4f}")
print(f"Precision: {rf_tuned_precision:.4f}")
print(f"Recall   : {rf_tuned_recall:.4f}")
print(f"F1-Score : {rf_tuned_f1:.4f}")
print(f"ROC-AUC  : {rf_tuned_auc:.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    rf_tuned_pred,
    target_names=["No Heart Disease", "Heart Disease"]
))
```

```python
# Final comparison of the best models

final_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Original Random Forest",
        "Tuned Random Forest"
    ],
    "Accuracy": [
        accuracy,
        rf_accuracy,
        rf_tuned_accuracy
    ],
    "Precision": [
        precision,
        rf_precision,
        rf_tuned_precision
    ],
    "Recall": [
        recall,
        rf_recall,
        rf_tuned_recall
    ],
    "F1-Score": [
        f1,
        rf_f1,
        rf_tuned_f1
    ],
    "ROC-AUC": [
        lr_auc,
        rf_auc,
        rf_tuned_auc
    ]
})

final_comparison = final_comparison.round(4)

print("Final Model Comparison")
print("=" * 70)
print(final_comparison.to_string(index=False))
```
